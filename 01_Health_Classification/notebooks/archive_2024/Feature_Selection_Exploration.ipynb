```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn import tree
```

```python
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
```

## Feature
 
- index : 인덱스
- name : 개인 이름
- birth : 생년월일
- bp_high, bp_low : 혈압 관련 지표
- chol_good, chol_bad : 콜레스테롤 지표
- iron_1v1, creatinine : 철분수치, 신장기능 지표
- urine_marker : 소변 마커
- sugar_1v1 : 혈당 수치
- lipid1v1, fat_content : 지질, 체지방 관련 지표
- oral_issues : 구강 질환
- stature_inch : 키
- mass_kg : 체중
- midsection_inch : 허리둘레
- immune_index : 면역력 상태 지표
- enzyme_1,2,3 : 특정 효소 수치
- vision_1, vision_r : 시력 관련 정보
- audio_1, audio_r : 청력 관련 정보
- health_status : 종합 건강 상태

```python
X_train=train_df.drop(["index","name","health_status"],axis=1)
X_test=test_df.drop(["index","name"],axis=1)
```

```python



# 'audio_l'과 'audio_r'에서 'N'을 1로, 'A'를 0으로 변경
X_train['audio_l'] = X_train['audio_l'].replace({'N': 0, 'A': 1})
X_train['audio_r'] = X_train['audio_r'].replace({'N': 0, 'A': 1})

X_test['audio_l'] = X_test['audio_l'].replace({'N': 0, 'A': 1})
X_test['audio_r'] = X_test['audio_r'].replace({'N': 0, 'A': 1})

# 'audio_l'과 'audio_r'의 OR 연산 결과로 'audio' 컬럼 생성
X_train['audio'] = X_train['audio_l'] | X_train['audio_r']
X_test['audio'] = X_test['audio_l'] | X_test['audio_r']


X_train['bmi']= X_train.mass_kg/((X_train.stature_inch+100)/100)**2
X_train['age']= 2024-X_train.birth+1
X_train['bp']= (X_train.bp_high-X_train.bp_low)
X_train['audio'] = X_train['audio_l'] | X_train['audio_r']

X_test['bmi']= X_test.mass_kg/((X_test.stature_inch+100)/100)**2
X_test['age']= 2024-X_test.birth+1
X_test['bp']= (X_test.bp_high-X_test.bp_low)
X_test['audio'] =X_test['audio_l'] | X_test['audio_r']

drop_colums=['stature_inch','mass_kg','birth','bp_high','bp_low','midsection_inch']
X_train=X_train.drop(drop_colums,axis=1)
X_test=X_test.drop(drop_colums,axis=1)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV 파일 읽기 (health_status와 bmi 컬럼이 포함된 데이터셋)
df = pd.read_csv('train.csv')

# health_status 값에 따른 bmi 분포
bmi_0 = X_train[df['health_status'] == 0]['enzyme_2']
bmi_1 = X_train[df['health_status'] == 1]['enzyme_2']

# 히스토그램 그리기
plt.figure(figsize=(10, 6))

plt.hist(bmi_0, bins=30, alpha=0.5, label='health_status = 0')
plt.hist(bmi_1, bins=30, alpha=0.5, label='health_status = 1')

plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.title('Distribution of BMI for health_status = 0 and health_status = 1')
plt.legend()

plt.show()
```
