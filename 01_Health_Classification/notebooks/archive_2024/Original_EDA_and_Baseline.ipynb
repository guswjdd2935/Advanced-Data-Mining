```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 로드
train_data = pd.read_csv('train.csv')
test_data = pd.read_csv('test.csv')

# 기존 인덱스를 열로 변환
train_data = train_data.reset_index()
test_data = test_data.reset_index()

# 분석에 사용할 열 선택 (name, health_status 제외)
train_excluded_columns = ['name', 'health_status']
test_excluded_columns = ['name']

train_features = train_data.drop(columns=train_excluded_columns).select_dtypes(include=['float64', 'int64'])
test_features = test_data.drop(columns=test_excluded_columns).select_dtypes(include=['float64', 'int64'])

# 1. Train과 Test의 개별 상관계수 계산 및 시각화
train_corr = train_features.corr()
test_corr = test_features.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(train_corr, cmap='Purples', annot=True, fmt='.1f', cbar=True, square=True, linewidths=0.5)
plt.title("Train")
plt.show()

plt.figure(figsize=(12, 10))
sns.heatmap(test_corr, cmap='Purples', annot=True, fmt='.1f', cbar=True, square=True, linewidths=0.5)
plt.title("Test Data Correlation Matrix")
plt.show()

# 2. Train + Test 데이터 결합 후 상관계수 계산 및 시각화
combined_data = pd.concat([train_features, test_features], axis=0)
combined_corr = combined_data.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(combined_corr, cmap='Purples', annot=True, fmt='.1f', cbar=True, square=True, linewidths=0.5)
plt.title("Combined Train + Test Correlation Matrix")
plt.show()
```

```python
import pandas as pd

# 데이터 로드
train_data = pd.read_csv('train.csv')
test_data = pd.read_csv('test.csv')

# 기존 인덱스를 열로 변환
train_data = train_data.reset_index()
test_data = test_data.reset_index()

# 분석에 사용할 열 선택 (name, health_status 제외)
train_excluded_columns = ['name', 'health_status']
test_excluded_columns = ['name']

train_features = train_data.drop(columns=train_excluded_columns).select_dtypes(include=['float64', 'int64'])
test_features = test_data.drop(columns=test_excluded_columns).select_dtypes(include=['float64', 'int64'])

# 데이터 크기에 따른 가중치 계산
train_weight = len(train_features) / (len(train_features) + len(test_features))
test_weight = len(test_features) / (len(train_features) + len(test_features))

# 각 데이터셋의 상관계수 계산
train_corr = train_features.corr()
test_corr = test_features.corr()

# 가중 평균 상관계수 계산
weighted_corr = train_corr * train_weight + test_corr * test_weight

# 결과 출력
print("Train Weight:", train_weight)
print("Test Weight:", test_weight)
print("Weighted Correlation Matrix:")
print(weighted_corr)

# 시각화
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 10))
sns.heatmap(weighted_corr, cmap='Purples', annot=True, fmt='.1f', cbar=True, square=True, linewidths=0.5)
plt.title("Weighted Correlation Matrix (Train + Test)")
plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn import tree
import numpy as np


#데이터셋 불러오기
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
drop=['audio_r','audio_l']
# 고윳값 Drop
X_train=train_df.drop(["index","name","health_status"],axis=1)
X_test=test_df.drop(["index","name"],axis=1)
y_train=train_df['health_status']

# 주요 피처 리스트
important_features = ['iron_lvl', 'stature_inch', 'fat_content', 'age_enzyme_ratio', 'Total_Cholesterol']

# 결과 확인
X_test
```

```python
X_train['audio_r']=X_train['audio_r'].astype('category')
X_train['audio_l']=X_train['audio_l'].astype('category')

X_test['audio_l']=X_test['audio_l'].astype('category')
X_test['audio_r']=X_test['audio_r'].astype('category')
```

```python
# k-fold 검증 설정
from sklearn.model_selection import KFold
from lightgbm import LGBMClassifier
from sklearn.metrics import roc_auc_score

kf = KFold(n_splits=5, shuffle=True, random_state=42)

# 결과 저장용 딕셔너리
feature_auc_scores = {}

# 각 Feature를 하나씩 Drop하면서 성능 측정
for feature in X_train.columns:
    X_temp = X_train.drop(columns=[feature])
    auc_scores = []
    
    for train_idx, val_idx in kf.split(X_temp):
        # Train/Validation 분리
        X_fold_train, X_fold_val = X_temp.iloc[train_idx], X_temp.iloc[val_idx]
        y_fold_train, y_fold_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
        
        # LightGBM 모델 정의 및 학습
        model = LGBMClassifier(random_state=42, n_estimators=100)
        model.fit(X_fold_train, y_fold_train)
        
        # AUC-ROC 점수 계산
        y_pred = model.predict_proba(X_fold_val)[:, 1]
        auc = roc_auc_score(y_fold_val, y_pred)
        auc_scores.append(auc)
    
    # Feature Drop 시 AUC 점수 저장
    feature_auc_scores[feature] = sum(auc_scores) / len(auc_scores)
```

```python
def create_interaction_features(df):
    df['Total_Cholesterol'] = df['chol_good'] + df['chol_bad']
    df['bmi'] = df['mass_kg'] / ((df['stature_inch'] + 100) / 100) ** 2
    df['bmi_iron'] = df['bmi'] * df['iron_lvl']
    df['Waist_to_Height_Ratio'] = df['midsection_inch']/ df['stature_inch']
    df['chol_good_iron'] = df['chol_good'] * df['iron_lvl']
    df['Enzyme_Avg'] = df[['enzyme_1', 'enzyme_2', 'enzyme_3']].mean(axis=1)
    df['lipid_enzyme3'] = df['lipid_lvl'] * df['enzyme_3']
    df['waist_to_age_ratio'] = df['midsection_inch'] / 2024-df['birth']
    df['iron_fat'] = df['iron_lvl'] * df['fat_content']
    df['stature_per_cholesterol'] = df['stature_inch'] / (df['Total_Cholesterol'] + 1e-6)
    df['iron_per_stature'] = df['iron_lvl'] / (df['stature_inch'] + 1e-6)
    df['iron_stature'] = df['iron_lvl'] * df['stature_inch']
    df['mass_iron_ratio'] = df['mass_kg'] / df['iron_lvl']
    
create_interaction_features(X_train)
create_interaction_features(X_test)
```
